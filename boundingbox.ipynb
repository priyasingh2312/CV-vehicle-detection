# Draw YOLO detections on an image

Fixed version: YOLO boxes from `box.data` are **x1, y1, x2, y2** (corners), not x, y, w, h.
The old code drew `(x, y)` to `(x + w, y + h)`, which gave wrong boxes. Class names are shown instead of numbers.

```python
import cv2
from ultralytics import YOLO

MODEL_PATH = "vehicle_model.pt"   # relative to this folder
IMAGE_PATH = "1.jpg"             # change to your image
CONF = 0.4

model = YOLO(MODEL_PATH)
image = cv2.imread(IMAGE_PATH)
if image is None:
    raise FileNotFoundError(f"Could not read {IMAGE_PATH}")

results = model(image, conf=CONF)

color = (0, 255, 0)
for result in results:
    for box in result.boxes:
        x1, y1, x2, y2 = map(int, box.xyxy[0].tolist())   # corner coordinates
        conf = float(box.conf[0])
        name = result.names[int(box.cls[0])]

        cv2.rectangle(image, (x1, y1), (x2, y2), color, 2)
        cv2.putText(image, f"{name} {conf:.2f}", (x1, max(y1 - 10, 15)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)

cv2.imshow("YOLO Detection", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
